# 🔍 DataFlow Sense: Translator SQL Hibrid (Deterministic + AI)

**DataFlow Sense** este un proiect dezvoltat pentru disciplina **Limbaje Formale și Tehnici de Compilare (LFT)**. Aplicația realizează traducerea interogărilor SQL în limbaj natural (Română), utilizând o arhitectură hibridă care combină analiza sintactică clasică cu procesarea limbajului natural (NLP).

---

## 🏗️ Arhitectura Sistemului

Proiectul integrează două metodologii distincte pentru procesarea limbajului:

1.  **Componenta Deterministă (Lex & Yacc):**
    * **Lexer (Flex):** Realizează analiza lexicală, identificând token-uri precum `SELECT`, `FROM`, `WHERE`, identificatori de tabele și operatori.
    * **Parser (Bison/Yacc):** Verifică gramatica SQL (LALR) și generează o traducere analitică (Arbore de Sintaxă Abstractă simulat).
    * **Backend:** Scris în **C**, compilat sub formă de executabil nativ.

2.  **Componenta Probabilistică (AI - Hugging Face):**
    * **Model:** `DistilBERT` (serializat local).
    * **Rol:** Clasifică tipul de interogare și oferă un scor de încredere, detectând contextul sau potențiale riscuri de securitate/performanță.

3.  **Interfața Utilizator (Streamlit):**
    * Dezvoltată în **Python**, servește drept punct de legătură între utilizator, modelul AI și executabilul C.

---

## 📂 Structura Proiectului

* `app.py` - Interfața Streamlit și logica de integrare Python-C.
* `sql.l` / `sql.y` - Sursa analizorului lexical și sintactic (Flex & Bison).
* `sql_model_local/` - Modelul AI pre-descărcat pentru funcționare offline.
* `Dockerfile` - Instrucțiunile de containerizare pentru deployment.
* `requirements.txt` - Dependențele librăriilor Python.
